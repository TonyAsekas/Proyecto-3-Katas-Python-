### Ejercicio 1:


```python
#Escribe una función que reciba una cadena de texto como parámetro 
# y devuelva un diccionario con las frecuencias de cada letra en la cadena. 
# Los espacios no deben ser considerados.

def contar_frecuencias (cadenas):
    
    frecuencias ={} # Inicializamos un diccionario vacio para acumular el recuento de cada caracter
    
    for caracter in cadenas: # Recorremos la cadena de texto caracter por caracter
        
        if caracter != " " : # Aqui vamos a filtrar los espacios
            
       
            if caracter in frecuencias:  # Vamos a comprobar si la letra ya está en el diccionario
                
                frecuencias[caracter] += 1 # Incrementamos en 1 el contador existente
            else:
                
                frecuencias[caracter] = 1 # Inicializamos la clave con valor 1
            
    return frecuencias # Ahora devolvemos el diccionario frecuencias modificado

# Ejemplo de uso 

cadena_texto = "Esta vida no es para corazones delicados" # Definimos la cadena de texto a contar

resultado = contar_frecuencias(cadena_texto) # Llamamos a la función pasándole la cadena y guardamos el diccionario devuelto

print(resultado) # Imprimimos el resultado en consola

```

### Ejercicio 2:

```python
# Dada una lista de números, obtén una nueva lista con el doble de cada valor. Usa la función map().

def obtener_dobles(numeros): # Creamos la función 
    
    resultado = list(map(lambda x: x * 2, numeros)) # Hacemos la operación para listar el doble de cada valor de numeros
    
    return resultado # Devolvemos el resultado

# Ejemplo de uso

numeros = [1,2,3,4,5,6,7,8,9] # Definimos la lista de numeros

resultado = obtener_dobles(numeros) # Llamamos a la funcion pasandole la lista y guardando la nueva

print(resultado) # Imprimimos la lista por consola
```

### Ejercicio 3:

```python
# Ejercicio 3: Escribe una función que tome una lista de palabras y una palabra objetivo como parámetros.
# La función debe devolver una lista con todas las palabras de la lista original que contengan la palabra objetivo.
 
def filtrar_palabra_objetivo(palabras, objetivo): # Creamos la función
    
    resultado = list(filter(lambda palabra: objetivo in palabra, palabras)) #Creamos la variables para buscar la palabra objetivo
    
    return resultado # Retornamos el resultado 

# Ejemplo de uso:

palabras = ["submarino", "casa", "subterráneo", "sol"] # Definimos la lista de palabras

objetivo = "sub" # Definimos la palabra objetivo

resultado = filtrar_palabra_objetivo(palabras, objetivo) # Llamamos a la funcion  guardando su resultado

print(resultado) # Imprimimos por consola
```

### Ejercicio 4:

```python
#Genera una función que calcule la diferencia entre los valores de dos listas. Usa la función map().

def diferencia_listas(lista1, lista2): #Creamos la función
    
    resultado = list(map(lambda x, y: x - y, lista1, lista2)) # Iniciamoss la variable para calcular la diferencia
    
    return resultado

# Ejemplo de uso:

lista1 = [10, 20, 30, 40] # Definimos la primera lista de números
lista2 = [2, 5, 10, 15]   # Definimos la segunda lista de números


resultado = diferencia_listas(lista1, lista2) # Llamamos a la función pasándole ambas listas y guardamos la diferencia

print(resultado) # Imprimimos la lista resultante por consola
```

### Ejercicio 5:

```python
# Escribe una función que tome una lista de números como parámetro 
# y un valor opcional nota_aprobado (por defecto 5).
# La función debe calcular la media de los números en la lista 
# y determinar si la media es mayor o igual que nota_aprobado. 
# Si es así, el estado será "aprobado"; de lo contrario, "suspenso". 
# La función debe devolver una tupla que contenga la media y el estado.

def evaluar_notas(notas , nota_aprobado = 5): # Iniciamos la función 
    
    media_notas = sum(notas) / len(notas) # Calculamos la media de las notas
    
    if(media_notas >= nota_aprobado): #Comprobamos si la media alcanza o supera la nota minima 
        
         estado = "Aprobado!" # Al igualar o superar la nota, le ponemos "Aprobado!"
         
    else:
        
        estado = "Suspenso..." # Al no llegar al minimo, el estadeo pasa a "Suspenso..."
        
    return media_notas, estado

# Ejemplo de uso:

mis_notas = [6, 7, 4, 8, 5] # Creamos una lista de notas

resultado = evaluar_notas(mis_notas) # Llamamos a la función usando la nota por defecto (5)

print(resultado) # Imprimimos la tupla (6.0, 'aprobado')
```

### Ejercicio 6:

```python
#Escribe una función que calcule el factorial de un número de manera recursiva.
# Ejercicio 6: Escribe una función que calcule el factorial de un número de manera recursiva.

def factorial(n): # Creamos la función recursiva
    
    if n == 0 or n == 1:  # Caso base: si n es 0 o 1, detenemos la recursividad
        return 1
    else:  # Caso recursivo: multiplicamos n por el factorial de (n - 1)
        return n * factorial(n - 1)


# Ejemplo de uso:

numero = 5 # Definimos el número para calcular su factorial

resultado = factorial(numero) # Llamamos a la función recursiva

print(f"El factorial de {numero} es: {resultado}") # Imprimimos el resultado (120)
    
```
