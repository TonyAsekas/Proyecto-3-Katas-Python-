## Strings


```python
texto = "Batman"

# Podemos indexar un string utilizando indices positicos o negativos para recorrerlo al derecho o al reves
print(texto[0])
print(texto[-1])
```

```python
#INDEXACION COMPLEJA
#string[start:stop:step]

#start -> indice de inicio
#stop -> indice final
#step -> saltos entre caracteres


personaje = "Gandalf"

#Extraer desde el 0 hasta el 3 -> el stop es exluido
print(personaje[0:3])

#Si no queremos indicar el star por obviedad podemos hacer
print(personaje[:3])

#Empezxamos desde el 3 y omitimos el final
print(personaje[3:])

#Quier oleer el string entero pero con un salto de dos en dos caracteres
print(personaje[::2])

#Innvertir un texo
print(personaje[::-1])
```

```python
# METODOS DE STRING

#Longitud
texto = "Ash"
print(len(texto))

#Lower -> Todas en minusculas
print(texto.lower())

#Upper -> Todas en mayusc ulas
print(texto.upper())

frase = "el caballero oscuro"

#Title -> Pone a cada substring la primera en mayuscula
print(frase.title())

#Capitalize -> Solo la primera mayuscula
print(frase.capitalize())

#Strip -> Elimina espacios iniciales y finales
frase_espacios = "    Hola, buenas tardes.    "
print(len(frase_espacios))
frase_espacios = frase_espacios.strip()
print(len(frase_espacios))

#Replace -> Reemplazazr subcadenas
frase_replace = "Buenas noches Gotham"
print(frase_replace.replace("Gotham", "Oviedo"))

#Find -> Encuentrea el primer indice del string o letra coincidente
frase = "Hola que tal"
print(frase.find("x"))

#Index -> HAce lo mismo que find, pero si no lo encuentra arroja un error
    #print(frase.index("x"))

#Join
superheroes = ["Batman", "Superman", "WonderWoman"]
print("_".join(superheroes))
```
