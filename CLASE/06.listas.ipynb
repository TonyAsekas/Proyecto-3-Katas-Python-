```python
lista =["String", 24, True, None, []]

animales = ["Perro", "Gato", "Ciervo", "Loro"]

#Longitud de una lista
print(len(animales))

#Acceder a los elementos de una lista
print(animales[3])
```

```python
# LISTAS ANIDADAS
peliculas = ["Terror",["The Evil Dead", "The Thing","The Exorcist"], "Sci-Fi",["2001", "Dune", "Gravity"]]
print(peliculas[1][0])

pelicula = "Jurassic Park"
print(list(pelicula))
```

```python
#INDEXACIÓN
peliculas = ["Terror",["The Evil Dead", "The Thing","The Exorcist"], "Sci-Fi",["2001", "Dune", "Gravity"]]

#[star:stop:step]

print(peliculas[2:4])
print(peliculas[2:])
print(peliculas[::2])
print(peliculas[::-1])
```

```python
# MÉTODOS DE LISTAS

# min max
nums = [1,2,45,34,2677,7564,345,234]
print(max(nums))
print(min(nums))

palabras = ["Hola", "Adiós", "Merienda", "Movil", "Raton", "Impresora"]
print(max(palabras))
print(min(palabras))

#sorted - sort -> sorted oprdena alfabeticamente o numericamente la lista a niveld e consulta y sort modifica la lista original
peliculas = ["The Evil Dead", "The Thing","The Exorcist","2001", "Dune", "Gravity"]
print(sorted(peliculas))
peliculas.sort()
print(peliculas)

nums = [1,2,45,34,2677,7564,345,234]
print(sorted(nums))

#count -> Hace recuento de cuantas veces aparece un elemento en la la lista
print(nums.count(34))

#reverse -> Modific a el orden al inverso de una lista
print(peliculas)
peliculas.reverse()
print(peliculas)
```

```python
# METODOS PARA MODIFICAR LOS ELEMENTOS DE LAS LISTAS

#append -> Añade un nuevo elemento al final de la lista
peliculas = ["The Evil Dead", "The Thing","The Exorcist","2001", "Dune", "Gravity"]
peliculas.append("El viaje de Chihiro")
print(peliculas)

#extend -> Añade al fiknal una lista nueva
pelis_extras = ["Spider-man", "Dracula", "Nosferatu"]
peliculas.extend(pelis_extras)
print(peliculas)

#Sustituir un elemento por otro
peliculas[3] = "Scream"
print(peliculas)

#insert -> Inserta un elementto en la lista en uina posicion especifica
peliculas.insert(1,"Jurassic Park")
print(peliculas)
```

```python
# MÉTODOS DE ELIMINACIÓN

peliculas = ["The Evil Dead", "The Thing","The Exorcist","2001", "Dune", "Gravity"]

#pop -> Por defecto elimina el ultimo elemento, pero si le pasamos un indice, elimina ese
peliculas.pop(1)
print(peliculas)

#remove -> Elimina el primer elemento coincidiente con el argumento, es decir, le pasamos el elemento para que lo borre
peliculas.remove("2001")
print(peliculas)

#clear -> Vacia toda la lista por completo
peliculas.clear()
print(peliculas)

```
