### Ejercicio 1:


```python
#Escribe una función que reciba una cadena de texto como parámetro 
# y devuelva un diccionario con las frecuencias de cada letra en la cadena. 
# Los espacios no deben ser considerados.

def contar_frecuencias (cadenas):
    
    frecuencias ={} # Inicializamos un diccionario vacio para acumular el recuento de cada caracter
    
    for caracter in cadenas: # Recorremos la cadena de texto caracter por caracter
        
        if caracter != " " : # Aqui vamos a filtrar los espacios
            
       
            if caracter in frecuencias:  # Vamos a comprobar si la letra ya está en el diccionario
                
                frecuencias[caracter] += 1 # Incrementamos en 1 el contador existente
            else:
                
                frecuencias[caracter] = 1 # Inicializamos la clave con valor 1
            
    return frecuencias # Ahora devolvemos el diccionario frecuencias modificado

# Ejemplo de uso 

cadena_texto = "Esta vida no es para corazones delicados" # Definimos la cadena de texto a contar

resultado = contar_frecuencias(cadena_texto) # Llamamos a la función pasándole la cadena y guardamos el diccionario devuelto

print(resultado) # Imprimimos el resultado en consola

```

### Ejercicio 5:

```python
# Escribe una función que tome una lista de números como parámetro 
# y un valor opcional nota_aprobado (por defecto 5).
# La función debe calcular la media de los números en la lista 
# y determinar si la media es mayor o igual que nota_aprobado. 
# Si es así, el estado será "aprobado"; de lo contrario, "suspenso". 
# La función debe devolver una tupla que contenga la media y el estado.

def evaluar_notas(notas , nota_aprobado = 5): # Iniciamos la función 
    
    media_notas = sum(notas) / len(notas) # Calculamos la media de las notas
    
    if(media_notas >= nota_aprobado): #Comprobamos si la media alcanza o supera la nota minima 
        
         estado = "Aprobado!" # Al igualar o superar la nota, le ponemos "Aprobado!"
         
    else:
        
        estado = "Suspenso..." # Al no llegar al minimo, el estadeo pasa a "Suspenso..."
        
    return media_notas, estado

# Ejemplo de uso:

mis_notas = [6, 7, 4, 8, 5] # Creamos una lista de notas

resultado = evaluar_notas(mis_notas) # Llamamos a la función usando la nota por defecto (5)

print(resultado) # Imprimimos la tupla (6.0, 'aprobado')
```

### Ejercicio 6:

```python
#Escribe una función que calcule el factorial de un número de manera recursiva.
# Ejercicio 6: Escribe una función que calcule el factorial de un número de manera recursiva.

def factorial(n): # Creamos la función recursiva
    
    if n == 0 or n == 1:  # Caso base: si n es 0 o 1, detenemos la recursividad
        return 1
    else:  # Caso recursivo: multiplicamos n por el factorial de (n - 1)
        return n * factorial(n - 1)


# Ejemplo de uso:

numero = 5 # Definimos el número para calcular su factorial

resultado = factorial(numero) # Llamamos a la función recursiva

print(f"El factorial de {numero} es: {resultado}") # Imprimimos el resultado (120)
    
```

```python
#Genera una función que, para un conjunto de caracteres, 
#devuelva una lista de tuplas con cada letra en mayúsculas y minúsculas.
#Las letras no pueden estar repetidas. Usa la función map().

def letras_mayusculas_minusculas(conjunto_caracteres): # Creamos la función para obtener las letras en mayúsculas y minúsculas
    
    letras_unicas = set(conjunto_caracteres) # Creamos un conjunto para eliminar las letras repetidas
    
    return list(map(lambda letra: (letra.upper(), letra.lower()), letras_unicas)) #Retornamos la lista de tuplas con las letras en mayúsculas y minúsculas

# Ejemplo de uso:

conjunto = "Hola, ¿cómo estás?"

print(letras_mayusculas_minusculas(conjunto))
```

```python
#Crea una función que tome una lista de dígitos y devuelva el número correspondiente.
# Por ejemplo, [5,7,2] corresponde al número 572.
# Usa la función reduce().

from functools import reduce

def construir_numero(lista_digitos): # Creamos la función para construir el número a partir de la lista de dígitos
    
    
    numero = reduce(lambda acc, d: acc * 10 + d, lista_digitos) # Usa reduce() con la lambda matemática sobre lista_digitos
    
    return numero # Retornamos el número construido

# Ejemplo de uso:
digitos = [5, 7, 2]
resultado = construir_numero(digitos)

print(f"La lista {digitos} corresponde al número: {resultado}")
```

```python
# Crea una función que busque y devuelva el primer elemento duplicado en una lista dada.

def primer_duplicado(lista): # Creamos la función para buscar el primer elemento duplicado en una lista
    elementos_vistos = set() # Creamos un conjunto para almacenar los elementos que ya hemos visto
    
    for elemento in lista: # Recorremos la lista elemento por elemento
        
        if elemento in elementos_vistos: # Comprobamos si el elemento ya está en el conjunto
            
            return elemento # Si es así, devolvemos el primer elemento duplicado encontrado
        
        elementos_vistos.add(elemento) # Si no, añadimos el elemento al conjunto de elementos vistos
    
    return None # Si no se encuentra ningún duplicado, devolvemos None

# Ejemplo de uso:
mi_lista = [1, 2, 3, 4, 5, 3, 6, 7] # Definimos la lista con elementos duplicados

resultadod = primer_duplicado(mi_lista) # Llamamos a la función pasándole la lista y guardamos el resultado

print(f"El primer elemento duplicado en la lista {mi_lista} es: {resultadod}") # Imprimimos el resultado por consola
```

### Ejercicio 29:

```python
#Crea una función que convierta una variable en una cadena de texto 
# y enmascare todos los caracteres con el carácter '#' excepto los últimos cuatro.

def enmascarar_variable(variable): # Creamos la función para enmascarar la variable
    
    variable_str = str(variable) # Convertimos la variable a cadena de texto
    
    longitud = len(variable_str) # Obtenemos la longitud de la cadena
    
    if longitud <= 4: # Comprobamos si la longitud es menor o igual a 4
        
        return variable_str # Si es así, devolvemos la cadena tal cual
    
    enmascarado = '#' * (longitud - 4) + variable_str[-4:] # Enmascaramos todos los caracteres excepto los últimos cuatro
    
    return enmascarado # Retornamos la cadena enmascarada

# Ejemplo de uso:

mi_variable = 1234567890 # Definimos la variable a enmascarar

resultado = enmascarar_variable(mi_variable) # Llamamos a la función pasándole la variable y guardamos el resultado

print(f"La variable enmascarada es: {resultado}") # Imprimimos el resultado por consola
```

```python
# Crea una función que determine si dos palabras son anagramas,
# es decir, si están formadas por las mismas letras pero en diferente orden.

def son_anagramas(palabra1, palabra2): # Creamos la función para determinar si dos palabras son anagramas
    
    # Eliminamos espacios y convertimos a minúsculas para una comparación justa
    palabra1 = palabra1.replace(" ", "").lower()
    palabra2 = palabra2.replace(" ", "").lower()
        
    return sorted(palabra1) == sorted(palabra2) # Comparamos las palabras ordenadas alfabéticamente

# Ejemplo de uso:
palabra1 = input("Introduce la primera palabra: ") # Pedimos al usuario la primera palabra
palabra2 = input("Introduce la segunda palabra: ") # Pedimos al usuario la segunda palabra

if son_anagramas(palabra1, palabra2): # Llamamos a la función y comprobamos si son anagramas
    
    print(f"'{palabra1}' y '{palabra2}' son anagramas.")
    
else:
    
    print(f"'{palabra1}' y '{palabra2}' NO son anagramas.")
```

```python
"""Crea la clase Arbol
Define un árbol genérico con un tronco y ramas como atributos.
Métodos disponibles: crecer_tronco, nueva_rama, crecer_ramas, quitar_rama, info_arbol.
Código a seguir:
Inicializar un árbol con un tronco de longitud 1 y una lista vacía de ramas.
Implementar el método crecer_tronco para aumentar la longitud del tronco en una unidad.
Implementar el método nueva_rama para agregar una nueva rama de longitud 1 a la lista de ramas.
Implementar el método crecer_ramas para aumentar en una unidad la longitud de todas las ramas existentes.
Implementar el método quitar_rama para eliminar una rama en una posición específica.
Implementar el método info_arbol para devolver información sobre la longitud del tronco, el número de ramas 
y sus longitudes.
Caso de uso:
        a. Crear un árbol.
        b. Hacer crecer el tronco una unidad.
        c. Añadir una nueva rama.
        d. Hacer crecer todas las ramas una unidad.
        e. Añadir dos nuevas ramas.
        f. Retirar la rama situada en la posición 2.
        g. Obtener información sobre el árbol."""
        
class Arbol: # Creamos la clase Arbol
    
    def __init__(self): # Inicializamos el árbol con un tronco de longitud 1 y una lista vacía de ramas
        
        self.tronco = 1
        self.ramas = []
        
    def crecer_tronco(self): # Método para hacer crecer el tronco en una unidad
        
        self.tronco += 1
        
    def nueva_rama(self): # Método para agregar una nueva rama de longitud 1 a la lista de ramas
        
        self.ramas.append(1)
        
    def crecer_ramas(self): # Método para hacer crecer todas las ramas en una unidad
        
        self.ramas = [rama + 1 for rama in self.ramas]
        
    def quitar_rama(self, posicion): # Método para eliminar una rama en una posición específica
        
        if 0 <= posicion < len(self.ramas):
            del self.ramas[posicion]
        else:
            print("Posición inválida. No se puede quitar la rama.")
            
    def info_arbol(self): # Método para devolver información sobre el árbol
        
        return {
            "longitud_tronco": self.tronco,
            "numero_ramas": len(self.ramas),
            "longitudes_ramas": self.ramas
        }
        
# Caso de uso:
mi_arbol = Arbol()           # a. Crear un árbol
mi_arbol.crecer_tronco()     # b. Hacer crecer el tronco una unidad
mi_arbol.nueva_rama()        # c. Añadir una nueva rama (ramas: [1])
mi_arbol.crecer_ramas()      # d. Hacer crecer todas las ramas una unidad (ramas: [2])
mi_arbol.nueva_rama()        # e. Añadir dos nuevas ramas
mi_arbol.nueva_rama()        #    (ramas: [2, 1, 1])
mi_arbol.quitar_rama(2)      # f. Retirar la rama en la posición 2 (ramas: [2, 1])

# g. Obtener información sobre el árbol
informacion = mi_arbol.info_arbol()
print(informacion)
```

```python
"""Crea la clase UsuarioBanco
Representa a un usuario de un banco con su nombre, saldo y si tiene o no cuenta corriente.
Métodos: retirar_dinero, transferir_dinero, agregar_dinero.
Código a seguir:
Inicializar un usuario con nombre, saldo y un indicador (True o False) de cuenta corriente.
Implementar retirar_dinero para sustraer dinero del saldo, lanzando un error si no es posible.
Implementar transferir_dinero para transferir dinero desde otro usuario, lanzando un error en caso de fallo.
Implementar agregar_dinero para aumentar el saldo del usuario.
Caso de uso:
        a. Crear dos usuarios: "Alicia" con saldo inicial de 100 y "Bob" con saldo inicial de 50,
        ambos con cuenta corriente.
        b. Agregar 20 unidades al saldo de Bob.
        c. Transferir 80 unidades de Bob a Alicia.
        d. Retirar 50 unidades del saldo de Alicia."""
        
class UsuarioBanco: # Creamos la clase
    
    def __init__(self, nombre, saldo, cuenta_corriente): # Inicializamos el usuario con nombre, saldo y cuenta corriente
        
        self.nombre = nombre
        self.saldo = saldo
        self.cuenta_corriente = cuenta_corriente
        
    def retirar_dinero(self, cantidad): # Método para retirar dinero del saldo
        
        if cantidad <= self.saldo: # Comprobamos si hay suficiente saldo
            
            self.saldo -= cantidad # Restamos la cantidad del saldo
            
        else:
            
            raise ValueError(f"Saldo insuficiente para retirar {cantidad}. Saldo actual: {self.saldo}") # Lanzamos un error si no hay suficiente saldo  
    
    def transferir_dinero(self, otro_usuario, cantidad): # Método para transferir dinero a otro usuario
        
        if cantidad <= self.saldo: # Comprobamos si hay suficiente saldo
            
            self.saldo -= cantidad # Restamos la cantidad del saldo del usuario actual
            otro_usuario.agregar_dinero(cantidad) # Agregamos la cantidad al saldo del otro usuario
            
        else:
            
            raise ValueError(f"Saldo insuficiente para transferir {cantidad}. Saldo actual: {self.saldo}") # Lanzamos un error si no hay suficiente saldo
            
            
    def agregar_dinero(self, cantidad): # Método para agregar dinero al saldo
        
        self.saldo += cantidad # Sumamos la cantidad al saldo del usuario
        
    def __str__(self): # Método para representar el objeto como cadena de texto
        
        return f"Usuario: {self.nombre}, Saldo: {self.saldo}, Cuenta Corriente: {self.cuenta_corriente}" # Retornamos la información del usuario como cadena de texto
    
    # Caso de uso:
 
 # 1. Creamos dos users
    
user1 = UsuarioBanco("Tony", 100, True) # Creamos usuario con nombre, saldo y cuenta corriente

user2 = UsuarioBanco("Sara", 50, True) # Creamos otro usuario con nombre, saldo y cuenta corriente

# 2. Agregamos 20 euros al saldo de Sara

user2.agregar_dinero(20)

# 3. Transferimos 70 euros de Sara a Tony

user2.transferir_dinero(user1, 70)

# 4. Retiramos 50 euros del saldo de Tony

user1.retirar_dinero(50)

# Imprimimos la información de ambos usuarios después de las operaciones

print(user1) # Imprimimos la información del primer usuario

print(user2) # Imprimimos la información del segundo usuario
```

```python
#Escribe una función que tome dos parámetros:
# figura (una cadena que puede ser "rectangulo", "circulo" o "triangulo")
# y datos (una tupla con los datos necesarios para calcular el área de la figura).

import math # Importamos la librería math para usar el valor de pi en el cálculo del área del círculo

def calcular_area(figura, datos): # Creamos la función para calcular el área de la figura
    
    if figura == "rectangulo": # Comprobamos si la figura es un rectángulo
        
        base, altura = datos # Desempaquetamos los datos de la tupla
        area = base * altura # Calculamos el área del rectángulo
        
    elif figura == "circulo": # Comprobamos si la figura es un círculo
        
        radio, = datos # Desempaquetamos los datos de la tupla
        area = math.pi * radio ** 2 # Calculamos el área del círculo
        
    elif figura == "triangulo": # Comprobamos si la figura es un triángulo
        
        base, altura = datos # Desempaquetamos los datos de la tupla
        area = (base * altura) / 2 # Calculamos el área del triángulo
        
    else:
        
        return "Figura no reconocida. Por favor, introduce 'rectangulo', 'circulo' o 'triangulo'." # Retornamos un mensaje de error si la figura no es reconocida
    
    return area # Retornamos el área calculada

# Ejemplo de uso:

figura_usuario = input("Introduce la figura (rectangulo, circulo, triangulo): ") # Pedimos al usuario que introduzca la figura

print("Introduce los datos necesarios para calcular el área:") # Pedimos al usuario que introduzca los datos necesarios

if figura_usuario == "rectangulo" or figura_usuario == "triangulo": # Comprobamos si la figura es un rectángulo o un triángulo
    
    base = float(input("Base: ")) # Pedimos la base
    
    altura = float(input("Altura: ")) # Pedimos la altura
   
    datos_usuario = (base, altura) # Creamos la tupla con los datos

elif figura_usuario == "circulo": # Comprobamos si la figura es un círculo
    
    radio = float(input("Radio: ")) # Pedimos el radio
    
    datos_usuario = (radio,) # Creamos la tupla con los datos
    
else:
    
   
    print("Figura no reconocida. Por favor, introduce 'rectangulo', 'circulo' o 'triangulo'.") # Mostramos un mensaje de error si la figura no es reconocida
    
    exit() # Salimos del programa

print(f"El área de {figura_usuario} es: {calcular_area(figura_usuario, datos_usuario)}") # Llamamos a la función y mostramos el resultado
```

```python
"""
Escribe un programa en Python que utilice condicionales para determinar el monto final de una compra en una tienda en línea,
después de aplicar un descuento. El programa debe:
    a. Solicitar al usuario el precio original de un artículo.
    b. Preguntar si tiene un cupón de descuento (respuesta sí o no).
    c. Si la respuesta es sí, solicitar el valor del cupón de descuento.
    d. Aplicar el descuento al precio original, siempre que el valor del cupón sea válido (mayor a cero).
    e. Mostrar el precio final de la compra, considerando o no el descuento.
    f. Usar estructuras de control de flujo (if, elif, else) para llevar a cabo las acciones.
"""

def calcular_precio_final(precio, tiene_cupon, valor_cupon=None): # Creamos la función para calcular el precio final de la compra

    if tiene_cupon: # Comprobamos si el usuario tiene un cupón de descuento

        if valor_cupon is not None and valor_cupon > 0: # Comprobamos si el valor del cupón es válido

            precio_final = precio - valor_cupon # Aplicamos el descuento al precio original

            if precio_final < 0: # Comprobamos si el precio final es menor a cero

                precio_final = 0 # Si es así, establecemos el precio final en cero
                
            else:
                print("Valor del cupón inválido. No se aplicará ningún descuento.") # Mostramos un mensaje de error si el valor del cupón no es válido
                
                precio_final = precio # El precio final será el precio original
                
        else:
                
                precio_final = precio # Si no tiene cupón, el precio final será el precio original

        return precio_final # Retornamos el precio final

# Ejemplo de uso:

precio_original = float(input("Introduce el precio original del artículo: ")) # Pedimos al usuario que introduzca el precio original del artículo

tiene_cupon = input("¿Tienes un cupón de descuento? (si/no): ").strip().lower() == "si" # Preguntamos si tiene un cupón de descuento

valor_cupon = None

if tiene_cupon:

    valor_cupon = float(input("Introduce el valor del cupón de descuento: ")) # Pedimos el valor del cupón de descuento

precio_final = calcular_precio_final(precio_original, tiene_cupon, valor_cupon)

print(f"El precio final de la compra es: {precio_final}") 

```
