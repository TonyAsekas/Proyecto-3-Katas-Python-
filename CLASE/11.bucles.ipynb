```python
dias_semana = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes","Sábado","Domingo"]

for dia in dias_semana:
    print(f'Hoy es: {dia}')
```

```python
nombres = ["laUra pEreZ","ÁlVarO RodRigUeZ","saRA heRedIA","daNiEL gArcIA"]

nombres_corregidos = []

for nombre in nombres :
    nombres_corregidos.append(nombre.title())
    
print(nombres_corregidos)
```

```python
numeros =[1,2,3,4,5,6,7,8,9]

total = 0

for numero in numeros: 
    total += numero
    
print(total)
```

```python
meses = {
    1: "enero",
    2: "febrero",
    3: "marzo",
    4: "abril",
    5: "mayo",
    6: "junio",
    7: "julio",
    8: "agosto",
    9: "septiembre",
    10: "octubre",
    11: "noviembre",
    12: "diciembre",
}

for numero, mes in meses.items():
    print(f'El mes {mes} tiene el número {numero}')
```

```python
dias = ["lunes","martes","miercoles","jueves","viernes","sabado","domingo"]

for i in  range(len(dias)):
    print(f'La posición {i} tiene el valor {dias[i]}')
```

```python
dias = ["lunes","martes","miercoles","jueves","viernes","sabado","domingo"]

for indice, dia in enumerate(dias):
    print(f'La posición {indice} tiene el valor {dia}')
```
