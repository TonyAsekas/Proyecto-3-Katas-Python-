```python
# LAS TUPLAS SON INMUTABLES

heroe = ("Spider-Man", "Peter Parker", 25, "Marvel")
print(len(heroe))
print(heroe.index("Peter Parker"))
print(heroe.count("Spider-Man"))
print(heroe[0])
print(heroe[1:3])

# Para crear una tupla, minimo tiene que haber una com a dntroi de los parentesis,
# aunque una tupla de un solo elemento,
# no tiene sentido, dado a que no podemos insertarle nada despues porque es inmutable

nueva_tupla=("Hola",)
print(type(nueva_tupla))

# Ejemplo uso TUPLA
config_server = ("localhost", 8080, True)
print(f"Conectado a {config_server[0]}:{config_server[1]}, SSL: {config_server[2]}")
```

```python
# TRUCO PARA MODIFICAR UNA TUPLA
config_server = ("localhost", 8080, True)
lista_config = list(config_server)
print(type(lista_config))

lista_config[0] = "Paco"
config_server = tuple(lista_config)


print(f"Conectado a {config_server[0]}:{config_server[1]}, SSL: {config_server[2]}")
```
