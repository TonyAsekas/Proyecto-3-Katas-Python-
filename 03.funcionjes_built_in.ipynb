```python
#TYPE

titulo = "Esto es un titulo"
numero = 25
num_decimal = 25.5
sin_valor = None
lista = [1,2,3]
tupla = (1,2,3)
diccionario = {"clave":"valor"}

print(type(titulo))
print(type(numero))
print(type(num_decimal))
print(type(sin_valor))
print(type(lista))
print(type(tupla))
print(type(diccionario))
```
