```python
                    #parametro
def saludar_con_nombre(nombre):
    print(f'Hola {nombre}')
    
                    #argumento
saludar_con_nombre("Tony")
saludar_con_nombre("Sara")
saludar_con_nombre("Koda")
```

```python
def sumar_dos_numeros(numeroA, numeroB):
    total = numeroA + numeroB
    print(total)
    
sumar_dos_numeros(5,6)
```

```python
def multiplicar_por_dos(numero):
    total  = numero * 2
    return total
    
resultado = multiplicar_por_dos(11) + multiplicar_por_dos(2)
print(resultado)
```

```python
def sumar(a, b):
    """
    Sumar dos numeros y retornar el resutlado total

    Args:
        numeroa (int, float): El primer numero de la suma
        numerob (int, float): El segundo numero de la suma

    Returns:
        int, float: Retorna el resultado de la suma
    """
    total = a + b
    return total
```

```python
# args -> parametro qwue recibe una cantidad indefinida/variable de valores

def mostrar_nombres(*args):
    for nombre in args:
        print("Los nombres son: ", args)
    
mostrar_nombres("Tony","Paco","Sara","Luis","Fredo")

def sumar_numeros(*args):
    total = 0
    for numero in args:
        total += numero
        print(total)
        
sumar_numeros(1,2,3,4,5,6,7,8,9)    
```

```python
#kwargs -> recibe una cantidad indefinida de clavbes con valores, generando asi un min i-diccionario

def mostrar_info (**kwargs):
    for clave, valor in kwargs.items():
        print(f'{clave}->{valor}')
    
mostrar_info(nombre="Ana", edad=20, ciudad="Madrid", profesion="Carpintera") 
mostrar_info(nombre="Pedro", edad=30, ciudad="Valencia", profesion="DJ") 
```

```python
# lambda -> Funciones cortas reutrilizables cn retorno implicito

def sumar (numa, numb):
    return numa + numb

resultado_suma = lambda numa, numb: numa + numb
print(resultado_suma(2,5))
```
