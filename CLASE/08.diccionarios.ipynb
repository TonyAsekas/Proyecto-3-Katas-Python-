## Diccionarios


```python
alumno = {
    "nombre":"Pepe",
    "edad": 32,
    "ciudad": "Granada"
}
print(alumno["nombre"])
alumno["matriculado"] = True
print(alumno)
alumno["nombre"] = "Tony"
print(alumno)

########

#Si intentamos acceder a una propiedad o clave sin que exista, nos arrojara un error
#print(alumno["altura"])

print(alumno.get("altura","no existe altura"))
print(alumno.get("nombre"))
```

```python
#METODOS DE DICCIONARIOS
alumno = {
    "nombre":"Pepe",
    "edad": 32,
    "ciudad": "Granada"
}

#Keys -> esto devuelve una vista de todas las claves
print(alumno.keys())

#Values -> esto devuelve una lista de los valores
print(alumno.values())

#Items -> Devuelve una vista de pares en tuplas de clave y valor
print(alumno.items())

#Upadte -> Actualizar un diccionario por otro diccionario
alumno.update({"altura": 1.76, "pelo": "rubio"})
print (alumno)

#Pop -> Le indicamos la clave y elimnia clave y valor
alumno.pop("pelo")
print(alumno)

#Copy ->HAce una copia superficial dde un diccionario
copia_alumno = alumno.copy()
copia_alumno["nombre"] = "Paco"
copia_alumno["edad"] = 30
print(copia_alumno)
print(alumno)

```
