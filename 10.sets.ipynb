```python
clasicos = {"Casablanca","Titanic","Forrest Gump","El Padrino","Gladiator"}
scifi ={"Matrix","Interstellar","Inception","Titanic","Avatar"}

# METODOS BASICOS

#Len
print(len (clasicos))

#Add -> Añadir elemento
scifi.add("Dune")
print(scifi)

#Upadte -> Añadir elementos en un listado
scifi.update(["2001", "Hail Mary"])
print(scifi)

#Remove -> Elimina unelemento especifico, si no da error
scifi.remove("Dune")
print(scifi)

#Discard -> Elimina unelemento espeficico, si no, no da error
scifi.discard("Dune")
```

```python
# METODOS ENTRE CONJUNTOS 
clasicos = {"Casablanca","Titanic","Forrest Gump","El Padrino","Gladiator"}
scifi ={"Matrix","Interstellar","Inception","Titanic","Avatar"}

#UNION -> | -> Unde dos conjuntos  y elimina duplicados
todos = clasicos.union(scifi)
todos = clasicos | scifi
print(todos)

#INTERSECCIÓN -> & -> Muestra elementos que aparecen en ambos conjuntos
comunes = clasicos.intersection(scifi)
comunes = clasicos & scifi
print(comunes)

#DIFERENCIA -> - ->  Muestra elementos que estan en uno pero no en el otro
solo_clasicos = clasicos.difference(scifi)
solo_clasicos = clasicos - scifi
print(solo_clasicos)

#DIFERENCIOA SIMETRICA -> ^ ->Elementos que estan en uno u otro conjunto pero no en ambos
unicos = clasicos.symmetric_difference(scifi)
unicos = clasicos ^ scifi
print(unicos)

#IsDisJoint -> Comprobar con un booleano si no tienen elementos en comun
print("¿Ciencia ficción y clásicos son únicas?",scifi.isdisjoint(clasicos))

#IsSubSet -> IsSuperSet -> Comprueba si un conjunto esta dentro de otro o lo contiene
miniset = {"Titanic", "El Padrino"}
print("¿El miniset está dentyro de clásicos?", miniset.issubset(clasicos))
print("¿Clásicos contiene el miniset?", clasicos.issuperset(miniset))
```
