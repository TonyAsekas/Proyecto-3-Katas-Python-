## Control Flow

```python
# break
# continue
# else
# pass
```

```python
# BREAK

numeros = [1,2,3,4,5,6,7]
for numero in numeros:
    print(numero)
    
    if numero == 3:
        break
    

alumnos =["Jorge","Pedro","Sara","Carla"]

for nombre in alumnos:
    print(f'Revisando: ', nombre)
    
    if (nombre == "Carla"):
        print("Carla encontrada")
        break
```

```python
# CONTINUE

numeros = [1,2,3,4,5,6,7]
for numero in numeros:
    if numero == 3:
        continue
    
    print(numero)
    
    
alumnos =["Jorge","Pedro","Sara","Byron","Carla"]
for alumno in alumnos:
    if(nombre == "Byron"):
        continue
print(f"Alumno normal: ", nombre)
```

```python
# ELSE
numeros = [1,2,3,4,5,6,7]

for numero in numeros :
    print(numero)
else:
    print("Bucle terminado")
```

```python
# PASS

for numero in range(5):
    pass

if(True):
    pass

alumnos =["Jorge","Pedro","Sara","Byron","Carla"]
for nombre in alumnos:
    if nombre == "Byron":
        pass
    else:
        print(nombre)

    
```
