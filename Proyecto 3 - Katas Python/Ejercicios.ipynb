### Ejercicio 1:


```python
#Escribe una función que reciba una cadena de texto como parámetro 
# y devuelva un diccionario con las frecuencias de cada letra en la cadena. 
# Los espacios no deben ser considerados.

def contar_frecuencias (cadenas):
    
    frecuencias ={} # Inicializamos un diccionario vacio para acumular el recuento de cada caracter
    
    for caracter in cadenas: # Recorremos la cadena de texto caracter por caracter
        
        if caracter != " " : # Aqui vamos a filtrar los espacios
            
       
            if caracter in frecuencias:  # Vamos a comprobar si la letra ya está en el diccionario
                
                frecuencias[caracter] += 1 # Incrementamos en 1 el contador existente
            else:
                
                frecuencias[caracter] = 1 # Inicializamos la clave con valor 1
            
    return frecuencias # Ahora devolvemos el diccionario frecuencias modificado

# Ejemplo de uso 

cadena_texto = "Esta vida no es para corazones delicados" # Definimos la cadena de texto a contar

resultado = contar_frecuencias(cadena_texto) # Llamamos a la función pasándole la cadena y guardamos el diccionario devuelto

print(resultado) # Imprimimos el resultado en consola

```

### Ejercicio 2:

```python
# Dada una lista de números, obtén una nueva lista con el doble de cada valor. Usa la función map().

def obtener_dobles(numeros): # Creamos la función 
    
    resultado = list(map(lambda x: x * 2, numeros)) # Hacemos la operación para listar el doble de cada valor de numeros
    
    return resultado # Devolvemos el resultado

# Ejemplo de uso

numeros = [1,2,3,4,5,6,7,8,9] # Definimos la lista de numeros

resultado = obtener_dobles(numeros) # Llamamos a la funcion pasandole la lista y guardando la nueva

print(resultado) # Imprimimos la lista por consola
```

### Ejercicio 3:

```python
# Ejercicio 3: Escribe una función que tome una lista de palabras y una palabra objetivo como parámetros.
# La función debe devolver una lista con todas las palabras de la lista original que contengan la palabra objetivo.
 
def filtrar_palabra_objetivo(palabras, objetivo): # Creamos la función
    
    resultado = list(filter(lambda palabra: objetivo in palabra, palabras)) #Creamos la variables para buscar la palabra objetivo
    
    return resultado # Retornamos el resultado 

# Ejemplo de uso:

palabras = ["submarino", "casa", "subterráneo", "sol"] # Definimos la lista de palabras

objetivo = "sub" # Definimos la palabra objetivo

resultado = filtrar_palabra_objetivo(palabras, objetivo) # Llamamos a la funcion  guardando su resultado

print(resultado) # Imprimimos por consola
```

### Ejercicio 4:

```python
#Genera una función que calcule la diferencia entre los valores de dos listas. Usa la función map().

def diferencia_listas(lista1, lista2): #Creamos la función
    
    resultado = list(map(lambda x, y: x - y, lista1, lista2)) # Iniciamoss la variable para calcular la diferencia
    
    return resultado

# Ejemplo de uso:

lista1 = [10, 20, 30, 40] # Definimos la primera lista de números
lista2 = [2, 5, 10, 15]   # Definimos la segunda lista de números


resultado = diferencia_listas(lista1, lista2) # Llamamos a la función pasándole ambas listas y guardamos la diferencia

print(resultado) # Imprimimos la lista resultante por consola
```

### Ejercicio 5:

```python
# Escribe una función que tome una lista de números como parámetro 
# y un valor opcional nota_aprobado (por defecto 5).
# La función debe calcular la media de los números en la lista 
# y determinar si la media es mayor o igual que nota_aprobado. 
# Si es así, el estado será "aprobado"; de lo contrario, "suspenso". 
# La función debe devolver una tupla que contenga la media y el estado.

def evaluar_notas(notas , nota_aprobado = 5): # Iniciamos la función 
    
    media_notas = sum(notas) / len(notas) # Calculamos la media de las notas
    
    if(media_notas >= nota_aprobado): #Comprobamos si la media alcanza o supera la nota minima 
        
         estado = "Aprobado!" # Al igualar o superar la nota, le ponemos "Aprobado!"
         
    else:
        
        estado = "Suspenso..." # Al no llegar al minimo, el estadeo pasa a "Suspenso..."
        
    return media_notas, estado

# Ejemplo de uso:

mis_notas = [6, 7, 4, 8, 5] # Creamos una lista de notas

resultado = evaluar_notas(mis_notas) # Llamamos a la función usando la nota por defecto (5)

print(resultado) # Imprimimos la tupla (6.0, 'aprobado')
```

### Ejercicio 6:

```python
#Escribe una función que calcule el factorial de un número de manera recursiva.
# Ejercicio 6: Escribe una función que calcule el factorial de un número de manera recursiva.

def factorial(n): # Creamos la función recursiva
    
    if n == 0 or n == 1:  # Caso base: si n es 0 o 1, detenemos la recursividad
        return 1
    else:  # Caso recursivo: multiplicamos n por el factorial de (n - 1)
        return n * factorial(n - 1)


# Ejemplo de uso:

numero = 5 # Definimos el número para calcular su factorial

resultado = factorial(numero) # Llamamos a la función recursiva

print(f"El factorial de {numero} es: {resultado}") # Imprimimos el resultado (120)
    
```

### Ejercicio  7:

```python
#Genera una función que convierta una lista de tuplas a una lista de strings.
#Usa la función map().

def tuplas_a_strings(lista_tuplas): # Creamos la función
    
    # Usamos f-string para acceder a cada elemento por su índice y darle formato
    resultado = list(map(lambda tupla: f"{tupla[0]} - {tupla[1]}", lista_tuplas))
    
    return resultado # Retornamos la lista de textos


# Ejemplo de uso con texto y números:

mis_tuplas = [("Juan", 25), ("María", 30), ("Carlos", 18)]

resultado = tuplas_a_strings(mis_tuplas)

print(resultado) # Salida: ['Juan - 25', 'María - 30', 'Carlos - 18']
```

### Ejercicio 9:

```python
#Escribe una función que tome una lista de nombres de mascotas como parámetro 
# y devuelva una nueva lista excluyendo ciertas mascotas prohibidas en España. 
# La lista de mascotas a excluir es 
# ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]. 
# Usa la función filter().

def filtrar_mascotas(lista_mascotas): # Creamos la función para filtrar las mascotas prohibidas
    prohibidas = ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"] # Hacemos una lista con las prohibidas
    
    # Filtramos conservando solo las mascotas que NOT IN prohibidas
    permitidas = list(filter(lambda mascota: mascota not in prohibidas, lista_mascotas))
    
    return permitidas # Retornamos la lista limpia


# Ejemplo de uso:

mis_mascotas = ["Perro", "Gato", "Conejo", "Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso", "Hámster"]

mascotas_legales = filtrar_mascotas(mis_mascotas)

print(mascotas_legales)
 
```

### Ejercicio 10:

```python
# Escribe una función que reciba una lista de números y calcule su promedio.
# Si la lista está vacía,
# lanza una excepción personalizada
# y maneja el error adecuadamente.

def calcular_promedio(lista_numeros): # Creamos la función para calcular el promedio

    if len(lista_numeros) == 0: # Comprobamos si la lista está vacía

        print("Error: la lista está vacía. No se puede calcular el promedio.") # Lanzamos la excepción personalizada

    else:

        promedio = sum(lista_numeros) / len(lista_numeros) # Calculamos el promedio

        return promedio # Retornamos el promedio



# Ejemplo de uso:

numeros = [1, 2, 3, 4, 5]

promedio = calcular_promedio(numeros)

print(f"El promedio es: {promedio}")
```

```python
#Genera una función que, al recibir una frase, 
# devuelva una lista con la longitud de cada palabra. 
# Usa la función map().

def longitud_palabras(frase): #Creamos la funcion
    
    palabras = frase.split()
    
    return list(map(len, palabras))

# Ejemplo de uso:

frase = "Hola, ¿cómo estás?"

print(longitud_palabras(frase))
```

### Ejercicio 13:

```python
#Genera una función que, para un conjunto de caracteres, 
#devuelva una lista de tuplas con cada letra en mayúsculas y minúsculas.
#Las letras no pueden estar repetidas. Usa la función map().

def letras_mayusculas_minusculas(conjunto_caracteres): # Creamos la función para obtener las letras en mayúsculas y minúsculas
    
    letras_unicas = set(conjunto_caracteres) # Creamos un conjunto para eliminar las letras repetidas
    
    return list(map(lambda letra: (letra.upper(), letra.lower()), letras_unicas)) #Retornamos la lista de tuplas con las letras en mayúsculas y minúsculas

# Ejemplo de uso:

conjunto = "Hola, ¿cómo estás?"

print(letras_mayusculas_minusculas(conjunto))
```

### Ejercicio 14:

```python
#Crea una función que retorne las palabras de una lista que comiencen con una letra en específico.
# Usa la función filter().

def filtrar_palabras_por_letra(lista_palabras, letra): # Creamos la función para filtrar las palabras por letra
    
    resultado = list(filter(lambda palabra: palabra[0].lower() == letra.lower(), lista_palabras)) # Filtramos las palabras que comiencen con la letra buscada
   
    return resultado

# Ejemplo de uso:
mis_palabras = ["Manzana", "Mesa", "Perro", "Gato", "mariposa", "Coche", "Mono"]
letra_buscada = "m"

resultado_filtrado = filtrar_palabras_por_letra(mis_palabras, letra_buscada)

print(f"Palabras que empiezan por '{letra_buscada}': {resultado_filtrado}")
```

### Ejercicio 16:

```python
#Escribe una función que tome una cadena de texto y un número entero n como parámetros 
#y devuelva una lista de todas las palabras que sean más largas que n. 
# Usa la función filter().

def palabras_mas_largas(cadena, n): # Creamos la función para filtrar las palabras más largas que n
    
    palabras = cadena.split() # Dividimos la cadena en palabras
    
    resultado = list(filter(lambda palabra: len(palabra) > n, palabras)) # Creamos la variable resultado para filtrar las palabras que sean más largas que n
    
    return resultado # Retornamos la lista de palabras más largas que n

# Ejemplo de uso:
cadena = "Si existe un Dios, el tendrá que rogarme a mi para que lo perdone" # Definimos la cadena de texto a analizar

n = 4 # Definimos el número entero n

resultado = palabras_mas_largas(cadena, n) # Llamamos a la función pasándole la cadena y el número n, y guardamos la lista devuelta

print(f"Palabras más largas que {n} caracteres: {resultado}") #Imprimimos el resultado en consola
```
